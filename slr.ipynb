```python
import numpy as np
```

```python
#create a dummy dataset
X = list(map(int, input("Enter sample input values:").split()))
Y = list(map(int, input("Enter sample output values:").split()))
```

```python
#check if input variables are of same length
if len(X) != len(Y):
    exit()
```

```python
#finding the mean value of the dataset
def num_mean(lists):
    total = 0
    count = 0
    for i in lists:
        total = total + i
        count = count + 1
    return total/count
        
X_mean = num_mean(X)
Y_mean = num_mean(Y)
```

```python
#convert lists to array
X = np.array(X)
Y = np.array(Y)
```

```python
#finding the slope
#Slope basically tells us the change in value of y, when x changes
#it tells us how steep the line is.
m = (np.sum((X - X_mean)*(Y - Y_mean)))/ (np.sum((X - X_mean)**2))
print("Slope:", m)
```

```python
#finding the intercept
#it is the value where a line crosses on axis
#for a simple, consider, value of y, when x = 0
c = Y_mean - (m * X_mean)
print("Intercept:", c)
```

```python
#write function to predict values
def predict(x):
    return (m * x) + c
```

```python
X = np.array(X)
Y_pred = predict(X)
print("Predicted values:", Y_pred)
```

```python
#plot line and regression line
import matplotlib.pyplot as plt
plt.scatter(X, Y)
plt.plot(X, Y_pred, label = "Regression line")
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Simple linear regression from scratch")
plt.show()
```

```python
#calculating the error
#here we find it using mean squared error (MSE)
error = np.mean((Y - Y_pred)**2)
print(error)
```
